Create the recommendation function

```python
def popularity_recommendation_by_genre(ratings, items, user_id, n=5):

    # Genre columns
    genre_cols = [
        'Action', 'Adventure', 'Animation', "Children's",
        'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy',
        'Film-Noir', 'Horror', 'Musical', 'Mystery', 'Romance',
        'Sci-Fi', 'Thriller', 'War', 'Western'
    ]

    # Find movies already rated by this user
    user_movies = ratings[
        ratings['user_id'] == user_id
    ]['movie_id'].unique()

    # Calculate movie popularity
    movie_popularity = ratings.groupby('movie_id').agg(
        rating_count=('rating', 'count'),
        average_rating=('rating', 'mean')
    ).reset_index()

    # Add movie title and genre information
    movie_popularity = movie_popularity.merge(
        items[['movie_id', 'movie_title'] + genre_cols],
        on='movie_id'
    )

    # Dictionary to store recommendations
    recommendations = {}

    # Process each genre
    for genre in genre_cols:

        # Select movies belonging to this genre
        genre_movies = movie_popularity[
            movie_popularity[genre] == 1
        ]

        # Remove movies already rated by the user
        genre_movies = genre_movies[
            ~genre_movies['movie_id'].isin(user_movies)
        ]

        # Sort by popularity
        genre_movies = genre_movies.sort_values(
            by=['rating_count', 'average_rating'],
            ascending=[False, False]
        )

        # Select top N movies
        recommendations[genre] = genre_movies[
            ['movie_id', 'movie_title', 'rating_count', 'average_rating']
        ].head(n)

    return recommendations
```

```python
recommendations = popularity_recommendation_by_genre(
    ratings,
    items,
    user_id=78,
    n=5
)
```

```python
for genre, movies in recommendations.items():

    print("\n====================================")
    print("Genre:", genre)
    print("====================================")

    print(movies.to_string(index=False))
```
