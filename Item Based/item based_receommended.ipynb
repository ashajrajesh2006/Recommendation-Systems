```python
import numpy as np

def predict(ratings, similarity, type='user'):

    ratings = np.asarray(ratings)

    if type == 'user':

        mean_user_rating = ratings.mean(axis=1)

        ratings_diff = ratings - mean_user_rating[:, np.newaxis]

        pred = (
            mean_user_rating[:, np.newaxis]
            + similarity.dot(ratings_diff)
            / np.abs(similarity).sum(axis=1)[:, np.newaxis]
        )

    elif type == 'item':

        pred = (
            ratings.dot(similarity)
            / np.abs(similarity).sum(axis=1)
        )

    return pred
```

Why? 
Different users rate differently.
Example
Person A gives mostly 5★
Person B gives mostly 3★
Instead of comparing absolute ratings, we compare how much they rate above or below their own average.
Returns the predicted ratings matrix (the version you posted has return pred,, which creates a one-element tuple; it's more common to write return pred unless you intentionally want a tuple).

```python
# prediction Table
user_prediction = predict(data_matrix, user_similarity, type='user')
item_prediction = predict(data_matrix, item_similarity, type='item') 
```

```python
def itembased(input_item, item_similarity, item_predictions,
              similar_item_count, thres):

    # 1. Convert item similarity into DataFrame
    item_sim_table = pd.DataFrame(item_similarity)

    # 2. Find similar movies
    similar_input_item = (
        item_sim_table[input_item]
        .sort_values(ascending=False)
        .head(similar_item_count)
        .index
    )

    # 3. Convert to list
    similar_item_input = list(similar_input_item)

    # 4. Find users who watched similar movies
    similar_item_userid_list = []

    for sim_item in similar_item_input:

        sim = list(
            ratings[ratings['movie_id'] == sim_item]['user_id']
        )

        similar_item_userid_list.append(sim)

    # 5. Flatten the list
    import itertools

    similar_item_userid_single_list = list(
        itertools.chain.from_iterable(similar_item_userid_list)
    )

    # 6. Get unique users
    unique_userid_similar_item = set(
        similar_item_userid_single_list
    )

    # 7. Find users who watched the input movie
    input_item_watched_userid = list(
        ratings[ratings['movie_id'] == input_item]['user_id']
    )

    # 8. Find movies watched by similar users
    recom = []

    for user in unique_userid_similar_item:

        movies_watched = list(
            ratings[ratings['user_id'] == user]['movie_id']
        )

        for movie in movies_watched:

            # Don't recommend the input movie
            if movie != input_item:

                # Avoid duplicates
                if movie not in recom:
                    recom.append(movie)

    # 9. Get prediction table
    highest_Rated = []

    user_pred = pd.DataFrame(item_predictions)

    # 10. Get predictions for input movie
    input_item_pre = pd.DataFrame(
        user_pred[input_item]
    )

    input_item_pred = input_item_pre.T

    # 11. Select movies above threshold
    for movie in recom:

        if movie in input_item_pred.columns:

            value = input_item_pred[movie].values[0]

            if value >= thres:
                highest_Rated.append(movie)

    return highest_Rated
```

# so the flow is this for the above function

      ↓
Find users who watched similar movies
      ↓
Find other movies those users watched
      ↓
Remove movies already watched
      ↓
Check predicted ratings
      ↓
Recommended movies
